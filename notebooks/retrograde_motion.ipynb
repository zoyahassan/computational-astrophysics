# Retrograde Motion

From Earth, Mars usually creeps eastward from one night to the next. Around the time Earth passes it, that drift reverses: Mars moves westward for a few weeks, then turns east again. The westward part is **retrograde motion**. Mars keeps orbiting the Sun in the same direction the whole time. What changes is the direction of our line of sight.

## Sidereal period

**Sidereal period (P):** how long it takes an object to complete one orbit relative to the stars.

Earth's sidereal period is

$$P_\oplus = 365.26\ \text{days}.$$

Mars's sidereal period is longer, about 687 days, because Mars orbits farther from the Sun.

## Synodic period

**Synodic period (S):** how long it takes an object to return to the same apparent position relative to the Earth and the Sun.

That is the time from one alignment of Sun–Earth–planet to the next. For Mars, this is the time between oppositions, when Mars is opposite the Sun in our sky.

## How P and S are related

$$
\frac{1}{S} =
\begin{cases}
\dfrac{1}{P} - \dfrac{1}{P_\oplus} & \text{(inferior)} \\[6pt]
\dfrac{1}{P_\oplus} - \dfrac{1}{P} & \text{(superior)}
\end{cases}
$$

- An **inferior** planet orbits inside Earth's orbit (Mercury, Venus), so it has the shorter sidereal period.
- A **superior** planet orbits outside Earth's orbit (Mars and the planets beyond), so Earth has the shorter sidereal period.

The two versions are the same idea: the synodic period comes from how fast one orbit gains on the other.

## Orbital speed

$$
v = \frac{d}{t}.
$$

Here **d** is the distance traveled in one orbit (the circumference) and **t** is the time for that orbit, which is the sidereal period **P**. So

$$
v = \frac{2\pi a}{P}.
$$

Earth, being closer to the Sun, has a shorter period and a higher orbital speed than Mars. Earth catches up to Mars, passes it, and pulls ahead. That overtaking is when Mars appears to move backward.


```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import numpy as np

from src.physics import AU, P_EARTH_DAYS, synodic_period_days

P_MARS = 686.98  # days
P_VENUS = 224.70  # days
A_MARS = 1.524  # AU

s_mars = synodic_period_days(P_MARS, superior=True)
s_venus = synodic_period_days(P_VENUS, superior=False)

def orbital_speed_au_per_day(radius_au, period_days):
    return 2 * np.pi * radius_au / period_days

v_earth = orbital_speed_au_per_day(1.0, P_EARTH_DAYS)
v_mars = orbital_speed_au_per_day(A_MARS, P_MARS)

print(f"Earth sidereal period P⊕ = {P_EARTH_DAYS:.2f} days")
print(f"Mars synodic period S    = {s_mars:.1f} days")
print(f"Venus synodic period S   = {s_venus:.1f} days")
print(f"Earth orbital speed      = {v_earth * AU / 86400 / 1000:.2f} km/s")
print(f"Mars orbital speed       = {v_mars * AU / 86400 / 1000:.2f} km/s")
```

## Why the lines of sight cross

Both planets orbit the Sun in the same direction. Earth moves faster. Around the time Earth passes Mars, the line from Earth to Mars swings backward across the sky.

While Earth is passing Mars, the lines of sight stop sweeping forward and swing back across one another. That backward swing is the retrograde part. Before and after it, the lines sweep steadily forward (prograde).
